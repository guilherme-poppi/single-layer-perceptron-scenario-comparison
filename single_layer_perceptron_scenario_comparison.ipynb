# Single-Layer Perceptron: Scenario Comparison

```python
import random
import pandas as pd
import math
import matplotlib.pyplot as plt
```

```python
scenario_1 = []
scenario_2 = []

def operations(x1, x2, x3, w1, w2, w3, bias):
    weighted_sum = (x1 * w1) + (x2 * w2) + (x3 * w3)
    z = weighted_sum + bias
    output = 1/(1+math.e**-z)
    return weighted_sum, z, output


for i in range(-25, 26):
    _, _, output_scenario_1 = operations(x1=i, x2=i, x3=i, w1=0.1, w2=0.1, w3=0.1, bias=1.5)
    scenario_1.append({'input_values_scenario_1': i, 'output': output_scenario_1})

    _, _, output_scenario_2 = operations(x1=i, x2=i, x3=0, w1=0.1, w2=0.1, w3=0.1, bias=1.5)
    scenario_2.append({'input_values_scenario_2': i, 'output': output_scenario_2})


df = pd.DataFrame(scenario_1)
df_2 = pd.DataFrame(scenario_2)

plt.plot(df['input_values_scenario_1'],
         df['output'],
         c='blue',
         label='Scenario 1')

plt.plot(df_2['input_values_scenario_2'],
         df_2['output'],
         c='red',
         label='Scenario 2')

plt.title('Perceptron Activation: Scenario Comparison', fontsize=14, pad=12)
plt.xlabel('Input', fontsize=13)
plt.ylabel('Sigmoid Output', fontsize=12)
plt.legend()
plt.show()
```
